```python
import pandas as pd
import numpy as np

from scipy.stats import ttest_ind

pd.set_option("display.max_columns", None)

df = pd.read_csv(
    "../data/insurance_data.txt",
    sep="|",
    low_memory=False
)

print(df.shape)
```

Feature Engineering

This section creates derived insurance risk metrics used throughout the hypothesis testing process.

```python
# loss ratio
df["LossRatio"] = (
    df["TotalClaims"] /
    df["TotalPremium"]
)

# margin
df["Margin"] = (
    df["TotalPremium"] -
    df["TotalClaims"]
)

# remove invalid values
df = df.replace([np.inf, -np.inf], np.nan)

df = df.dropna(
    subset=["LossRatio", "Margin"]
)

df.head()
```

Hypothesis 1: Province Risk Difference
Null Hypothesis (H₀)

There is no significant difference in insurance risk between Gauteng and Western Cape.

Alternative Hypothesis (H₁)

There is a significant difference in insurance risk between Gauteng and Western Cape.

KPI Used:

Loss Ratio

```python
gauteng = df[
    df["Province"] == "Gauteng"
]["LossRatio"]

western_cape = df[
    df["Province"] == "Western Cape"
]["LossRatio"]

stat, p_value = run_ttest(
    gauteng,
    western_cape
)

print("P-value:", p_value)

province_p = p_value
```

Decision Rule:
Reject H₀ if p < 0.05.

Hypothesis 2: Zip Code Risk Difference
H₀

There is no significant risk difference between selected postal codes.

KPI

Loss Ratio

```python
zip1 = df[
    df["PostalCode"] == 2000
]["LossRatio"]

zip2 = df[
    df["PostalCode"] == 8001
]["LossRatio"]

stat, p_value = run_ttest(
    zip1,
    zip2
)

print("P-value:", p_value)

zip_p = p_value
```

Hypothesis 3: Margin Difference by Postal Code
H₀

There is no significant margin difference between selected postal codes.

```python
margin_zip1 = df[
    df["PostalCode"] == 2000
]["Margin"]

margin_zip2 = df[
    df["PostalCode"] == 8001
]["Margin"]

stat, p_value = run_ttest(
    margin_zip1,
    margin_zip2
)

print("P-value:", p_value)

margin_p = p_value
```

Hypothesis 4: Gender Risk Difference
H₀

There is no significant insurance risk difference between women and men.

KPI

Loss Ratio

```python
male = df[
    df["Gender"] == "Male"
]["LossRatio"]

female = df[
    df["Gender"] == "Female"
]["LossRatio"]

stat, p_value = run_ttest(
    male,
    female
)

print("P-value:", p_value)

gender_p = p_value
```

Results Summary

```python
results = pd.DataFrame({
    "Hypothesis": [
        "Province Risk Difference",
        "Zip Code Risk Difference",
        "Margin Difference",
        "Gender Risk Difference"
    ],
    "Test": [
        "T-Test",
        "T-Test",
        "T-Test",
        "T-Test"
    ],
    "P-Value": [
        province_p,
        zip_p,
        margin_p,
        gender_p
    ]
})

results["Decision"] = np.where(
    results["P-Value"] < 0.05,
    "Reject H0",
    "Fail to Reject H0"
)

results
```
