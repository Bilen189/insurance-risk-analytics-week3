Insurance Risk Analytics — Hypothesis Testing
Business Objective

The objective of this analysis is to statistically validate whether significant insurance risk differences exist across provinces, postal codes, and gender groups.

The findings will support:

Risk-based pricing
Geographic segmentation
Customer targeting
Portfolio optimization
Marketing strategy decisions

Risk is evaluated using:

Loss Ratio
Margin
Claim behavior

Statistical hypothesis testing enables AlphaCare Insurance Solutions (ACIS) to make evidence-driven pricing decisions rather than relying on intuition.

```python
import pandas as pd
import numpy as np

from scipy.stats import ttest_ind

pd.set_option("display.max_columns", None)

df = pd.read_csv(
    "../data/insurance_data.txt",
    sep="|",
    low_memory=False
)

print(df.shape)
```

Feature Engineering

This section creates derived insurance risk metrics used throughout the hypothesis testing process.

```python
# loss ratio
df["LossRatio"] = (
    df["TotalClaims"] /
    df["TotalPremium"]
)

# margin
df["Margin"] = (
    df["TotalPremium"] -
    df["TotalClaims"]
)

# remove invalid values
df = df.replace([np.inf, -np.inf], np.nan)

df = df.dropna(
    subset=["LossRatio", "Margin"]
)

df.head()
```
