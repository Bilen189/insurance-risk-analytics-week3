```python
import pandas as pd

df = pd.read_csv("../data/insurance.csv")
df.head()
```

```python
df.info()
df.describe()
df.isnull().sum()
```

```python
df["risk_score"] = df["charges"]
```

```python
df["charges"].mean()
```

```python
df.groupby("smoker")["charges"].mean()
```

```python
df.groupby("sex")["charges"].mean()
```

```python
df.groupby("region")["charges"].mean()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(data=df, x="bmi", y="charges")
plt.show()
```

```python
sns.scatterplot(data=df, x="age", y="charges")
plt.show()
```

```python
sns.boxplot(data=df, x="smoker", y="charges")
plt.title("Charges: Smokers vs Non-Smokers")
plt.show()
```

```python
sns.scatterplot(data=df, x="bmi", y="charges", hue="smoker")
plt.title("BMI vs Insurance Charges")
plt.show()
```

```python
sns.barplot(data=df, x="region", y="charges")
plt.title("Average Charges by Region")
plt.show()
```

This dataset represents medical insurance pricing risk. Charges are used as a proxy for claim severity and overall insurance risk exposure.
