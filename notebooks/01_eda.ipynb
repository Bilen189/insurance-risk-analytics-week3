Insurance Risk Analytics: Exploratory Data Analysis
Business Understanding

AlphaCare Insurance Solutions (ACIS) aims to improve insurance pricing and risk segmentation using historical insurance policy and claims data.

The objective of this analysis is to:

Understand claim behavior across customer groups
Identify high-risk segments
Analyze profitability using Loss Ratio
Explore geographic and vehicle-related risk patterns
Support future predictive modeling and pricing optimization

Key metrics analyzed include:

TotalPremium
TotalClaims
Loss Ratio
Margin

Insurance Metrics

Two important business metrics were created:

Loss Ratio = TotalClaims / TotalPremium
Margin = TotalPremium - TotalClaims

These metrics help evaluate portfolio profitability and risk.

```python
df["LossRatio"] = (
    df["TotalClaims"] / df["TotalPremium"]
)

df["Margin"] = (
    df["TotalPremium"] - df["TotalClaims"]
)

df[["LossRatio", "Margin"]].head()
```

```python
df["LossRatio"] = np.where(
    df["TotalPremium"] > 0,
    df["TotalClaims"] / df["TotalPremium"],
    0
)

df["LossRatio"].describe()
```

```python
overall_loss_ratio = (
    df["TotalClaims"].sum() /
    df["TotalPremium"].sum()
)

print("Overall Loss Ratio:", overall_loss_ratio)
```

```python
province_risk = df.groupby(
    "Province"
)["LossRatio"].mean().sort_values(
    ascending=False
)

province_risk
```

```python
gender_risk = df.groupby(
    "Gender"
)["LossRatio"].mean()

gender_risk
```

```python
vehicle_risk = df.groupby(
    "VehicleType"
)["LossRatio"].mean().sort_values(
    ascending=False
)

vehicle_risk.head(10)
```

Correlation Analysis

Correlation analysis identifies relationships between numerical variables.

```python
numeric_cols = [
    "TotalPremium",
    "TotalClaims",
    "CustomValueEstimate",
    "LossRatio",
    "Margin"
]

correlation_matrix = df[numeric_cols].corr()

correlation_matrix
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Matrix")

plt.show()
```

Outlier Detection

Boxplots help identify extreme values that may skew analysis.

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    x=df["TotalClaims"]
)

plt.title("Boxplot of Total Claims")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    x=df["CustomValueEstimate"]
)

plt.title("Boxplot of Custom Value Estimate")

plt.show()
```

```python
df["TransactionMonth"] = pd.to_datetime(
    df["TransactionMonth"]
)

monthly_claims = df.groupby(
    df["TransactionMonth"].dt.to_period("M")
)["TotalClaims"].sum()

monthly_claims.plot(
    figsize=(12,5)
)

plt.title("Monthly Total Claims Trend")
plt.ylabel("Total Claims")
plt.show()
```

```python
plt.figure(figsize=(12,6))

province_risk.plot(
    kind="bar"
)

plt.title("Average Loss Ratio by Province")
plt.ylabel("Loss Ratio")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x="TotalPremium",
    y="TotalClaims"
)

plt.title(
    "Total Premium vs Total Claims"
)

plt.show()
```

```python
plt.figure(figsize=(6,5))

gender_risk.plot(
    kind="bar"
)

plt.title(
    "Average Loss Ratio by Gender"
)

plt.ylabel("Loss Ratio")

plt.show()
```

```python
vehicle_risk.head(10).plot(
    kind="bar",
    figsize=(12,6)
)

plt.title(
    "Top 10 Vehicle Types by Loss Ratio"
)

plt.ylabel("Loss Ratio")

plt.show()
```

Key Business Insights
Geographic Risk Patterns

Gauteng and Limpopo provinces demonstrated the highest average loss ratios, indicating higher insurance risk and reduced profitability in these regions.

Gender Risk Analysis

Female policyholders showed slightly higher average loss ratios compared to male policyholders.

Vehicle Risk Analysis

Heavy commercial vehicles exhibited the highest loss ratios, suggesting greater claim severity and operational risk.

Claims Trend

Total claims increased significantly during 2015, indicating possible changes in customer behavior, economic conditions, or operational exposure.

Profitability Insight

The relationship between TotalPremium and TotalClaims suggests that some high-premium policies still generate substantial claims exposure, emphasizing the importance of accurate risk-based pricing.
