```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

df = pd.read_csv(
    "../data/insurance_data.txt",
    sep="|",
    encoding="latin1"
)

df.head()
```

```python
missing_values = df.isnull().sum()

missing_values[missing_values > 0]
```

```python
print(df.columns)
```

```python
df.isnull().sum()
```

```python
df["LossRatio"] = np.where(
    df["TotalPremium"] > 0,
    df["TotalClaims"] / df["TotalPremium"],
    0
)

df["LossRatio"].describe()
```

```python
overall_loss_ratio = (
    df["TotalClaims"].sum() /
    df["TotalPremium"].sum()
)

print("Overall Loss Ratio:", overall_loss_ratio)
```

```python
province_risk = df.groupby(
    "Province"
)["LossRatio"].mean().sort_values(
    ascending=False
)

province_risk
```

```python
gender_risk = df.groupby(
    "Gender"
)["LossRatio"].mean()

gender_risk
```

```python
vehicle_risk = df.groupby(
    "VehicleType"
)["LossRatio"].mean().sort_values(
    ascending=False
)

vehicle_risk.head(10)
```

```python
df["TransactionMonth"] = pd.to_datetime(
    df["TransactionMonth"]
)

monthly_claims = df.groupby(
    df["TransactionMonth"].dt.to_period("M")
)["TotalClaims"].sum()

monthly_claims.plot(
    figsize=(12,5)
)

plt.title("Monthly Total Claims Trend")
plt.ylabel("Total Claims")
plt.show()
```

```python
plt.figure(figsize=(12,6))

province_risk.plot(
    kind="bar"
)

plt.title("Average Loss Ratio by Province")
plt.ylabel("Loss Ratio")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x="TotalPremium",
    y="TotalClaims"
)

plt.title(
    "Total Premium vs Total Claims"
)

plt.show()
```

```python
plt.figure(figsize=(6,5))

gender_risk.plot(
    kind="bar"
)

plt.title(
    "Average Loss Ratio by Gender"
)

plt.ylabel("Loss Ratio")

plt.show()
```

```python
vehicle_risk.head(10).plot(
    kind="bar",
    figsize=(12,6)
)

plt.title(
    "Top 10 Vehicle Types by Loss Ratio"
)

plt.ylabel("Loss Ratio")

plt.show()
```

### Insight
Gauteng province demonstrates higher average loss ratios compared to other provinces, suggesting elevated insurance risk and potentially higher claim frequency.
